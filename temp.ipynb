```python
import os
os.environ["HF_HOME"] = "data_/hf"
os.environ["HF_DATASETS_OFFLINE"] = "1"
import json
from datasets import load_dataset, disable_caching
disable_caching()
```

```python
splits = ["train", "val", "test"]
d = {}
for split in splits:
    with open(f"data_/hf/hub/datasets--cilabuniba--artseek-licn-data/snapshots/febde3ca54032ca61d058232396d175e60cfa137/{split}.json", "r") as f:
        d[split] = json.load(f)
d["train"].keys()
```

```python
tasks = ["artist", "tag", "genre", "style", "media"]
class_lookups = {task: {} for task in tasks}
for split in splits:
    for task in tasks:
        class_lookups[task].update(d[split][task])
```

```python
with open("class_lookups.json", "w") as f:
    json.dump(class_lookups, f)
```

```python
from qdrant_client import QdrantClient, models

client = QdrantClient(url="http://localhost", prefer_grpc=True)
collection_name = "wikifragments-visual-arts-embeds"
```

```python
client.get_collection(collection_name)
```

```python
client.delete_collection(collection_name)
client.create_collection(
    collection_name=collection_name,
    on_disk_payload=True,
    vectors_config={
        "pooled": models.VectorParams(
            size=128,
            distance=models.Distance.COSINE,
            on_disk=True,
            datatype=models.Datatype.FLOAT16,
            multivector_config=models.MultiVectorConfig(
                comparator=models.MultiVectorComparator.MAX_SIM
            ),
            quantization_config=models.BinaryQuantization(
                binary=models.BinaryQuantizationConfig(always_ram=True),
            ),
        ),
        "full": models.VectorParams(
            size=128,
            distance=models.Distance.COSINE,
            on_disk=True,
            datatype=models.Datatype.FLOAT16,
            multivector_config=models.MultiVectorConfig(
                comparator=models.MultiVectorComparator.MAX_SIM
            ),
            quantization_config=models.BinaryQuantization(
                binary=models.BinaryQuantizationConfig(),
            ),
            hnsw_config=models.HnswConfigDiff(m=0),
        ),
    },
    optimizers_config=models.OptimizersConfigDiff(indexing_threshold=0),
    shard_number=4,
)
```

```python
example = ds["train"][0]
examples = {
    k: [v] for k, v in ds["train"][0].items()
}
idxs = [0]
```

```python
points = [
    models.PointStruct(
        id=id,
        vector={"full": vector, "pooled": pooled_vector},
        payload={"idx": idx},
    )
    for id, pooled_vector, vector, idx in zip(
        examples["id"],
        examples["pooled_embeddings"],
        examples["full_embeddings"],
        idxs,
    )
]
print(points[0].vector)
client.upload_points(
    collection_name=collection_name,
    points=points,
)
```
